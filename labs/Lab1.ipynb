**3. Generating a sinusoid**

3.1. Create a sinusoid of the same duration than the reference signal and same frequency than the fundamental frequency. Plot it along with the audio signal of the reference recording. Plot the same time segment that 2.3. Note that the sampling rate should be the same for both signals. Try to find the values of Amplitude and initial Phase that make the two signal segments plotted to match as close as possible (do not attempt to imitate the whole signal). Matching the phase is a bit tricky.



```python
import numpy as np

# parameters
fs = fs              # same sampling rate as the reference (requested)
f0 = 440.5           # frequency from 2.1
A = 0.5              # adjust amplitude to match 
phi = np.pi/4             
t = np.arange(0, len(ref)) / fs   # same duration as reference

# create sinusoid
x = A * np.cos(2 * np.pi * f0 * t + phi) # 2pif = w

# plot both signals over the same 3-period as in 2.3
t_start = 0.2
t_end = t_start + 3 * (1 / f0)

plot_signals([ref, x], fs, t_start=t_start, t_end=t_end, name=['reference signal', 'generated sinusoid'])
```

3.2. What are the main differences between the two signals.

We can see that the reference signal (blue) is more complex, while the generated sinusoid (orange) is smooth and perfectly periodic. Moreover, the reference has small variations and additional oscillations because it contains many harmonics, and obviously a decaying amplitude (because it’s a real sound produced by an instrument).
The sinusoid only represents the fundamental frequency, with constant amplitude and no harmonics or decay.
Therefore, the sinusoid only approximates one simple component of the real sound.

3.3. Listen to the synthesized signal usign the Audio widget. Sounds natural?

The synthesized signal roughly imitates the “La” sound, but it maintains the same intensity throughout. What I mean with that; the original sound starts strong and gradually decreases in amplitude over time, producing a natural decay. Because of this, the pure sinusoid sounds artificial, it lacks the variations in amplitude and the harmonics that make real instrument sounds richer and more natural. So, in general, it kind of imitates the sound, but it's way far from sounding as natural as a real instrument.
